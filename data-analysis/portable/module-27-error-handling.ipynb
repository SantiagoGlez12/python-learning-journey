## 2. Abrir test.txt sin controlar el error
#### Intento abrir un archivo que no existe, sin utilizar try ni except.
#### Esta celda debe mostrar FileNotFoundError. El error es parte del ejercicio, no significa que el Notebook esté mal.

```python
# Ejemplo SIN control de errores (fallará a propósito).
archivo = open("test.txt", "r")
print(archivo.read())
archivo.close()
```

## 3. Manejar el error con try y except
#### Ahora repito el ejercicio, pero utilizo try para intentar abrir el archivo y except para mostrar un mensaje comprensible si no existe.

```python
try:
    archivo = open("test.txt", "r")
    print(archivo.read())
    archivo.close()
except FileNotFoundError as error:
    print("No se pudo abrir el archivo test.txt.")
    print("Detalle del error:", error)
```

## 4. Generar una excepción con raise
#### raise sirve para generar una excepción. Compruebo si existe el archivo y, como no está, genero el error con un mensaje escrito por mí.

```python
try:
    if not os.path.exists("test.txt"):
        raise FileNotFoundError("El archivo test.txt no existe en la carpeta de práctica.")
    print("El archivo sí existe.")
except FileNotFoundError as error:
    print("Error generado:", error)
```

## 5. Utilizar try, except, else y finally
#### - try: intenta abrir el archivo.
#### - except: se ejecuta si ocurre el error.
#### - else: se ejecuta solamente si no hubo error.
#### - finally: se ejecuta siempre.
#### En este caso, como test.txt no existe, se ejecutan except y finally.

```python
try:
    archivo = open("test.txt", "r")
    contenido = archivo.read()
    archivo.close()
except FileNotFoundError:
    print("EXCEPT: No encontré test.txt.")
else:
    print("ELSE: El archivo se abrió correctamente.")
    print(contenido)
finally:
    print("FINALLY: Terminé la búsqueda del archivo.")
```

## 6. Operación matemática con un error
#### Sumo cuatro números y después intento dividir la suma entre cero. Esto produce ZeroDivisionError.
#### Esta celda también fallará a propósito. Se puede ejecutar sola y continuar después con la siguiente celda.

```python
numero1 = 10
numero2 = 20
numero3 = 30
numero4 = 40

suma = numero1 + numero2 + numero3 + numero4
divisor = 0
resultado = suma / divisor

print("Resultado:", resultado)
```

## 7. Identificar el problema con mensajes y try/except
#### Antes de hacer la división, imprimo los valores para ver qué está pasando. Con try/except puedo mostrar un mensaje en lugar de detener el programa.

```python
numero1 = 10
numero2 = 20
numero3 = 30
numero4 = 40

suma = numero1 + numero2 + numero3 + numero4
divisor = 0

print("Suma:", suma)
print("Divisor:", divisor)

try:
    resultado = suma / divisor
    print("Resultado:", resultado)
except ZeroDivisionError:
    print("Error: no se puede dividir entre cero.")
```

## 8. Corregir el error
#### En este ejemplo cambio el divisor de 0 a 2. Ahora el programa puede calcular el resultado sin problema.

```python
numero1 = 10
numero2 = 20
numero3 = 30
numero4 = 40

suma = numero1 + numero2 + numero3 + numero4
divisor = 2
resultado = suma / divisor

print("Suma:", suma)
print("Divisor:", divisor)
print("Resultado corregido:", resultado)
```

## 9. Debugging paso a paso
#### - El depurador se detiene antes de ejecutar la división.
#### - Se observa que la suma es 100 y el divisor es cero.
#### - Al avanzar aparece la excepción ZeroDivisionError.

```python
def calcular_operacion():
    numero1 = 10
    numero2 = 20
    numero3 = 30
    numero4 = 40
    suma = numero1 + numero2 + numero3 + numero4
    divisor = 0
    resultado = suma / divisor  # Esta linea produce el error.
    return resultado

print("Resultado:", calcular_operacion())
```
