## Descargar el archivo CSV
#### Descarda del archivo desde Google Drive a carpeta Downloads de mi equipo

## Leer el archivo CSV
#### Uso pd.read_csv() para cargar el archivo en una variable llamada datos. Después reviso cuántos registros tiene y cuáles son sus columnas.

```python
datos = pd.read_csv(archivo)
print("Cantidad de registros:", len(datos))
print("Columnas del archivo:")
print(datos.columns.tolist())
```

## Mostrar las primeras 50 filas
#### Con head(50) obtengo las primeras 50 filas del archivo y con print() las muestro en pantalla.

```python
print(datos.head(50).to_string(index=False))
```

## Calcular la media, el mínimo y el máximo de las alturas

#### Para este ejercicio uso las columnas en centímetros. La media se calcula con la función importada mean, y para los valores más bajos y altos utilizo min y max.

##### (Estos cálculos usan los 199 registros del archivo, no solamente los primeros 50.)

```python
# Alturas de hombres.
media_hombres = mean(datos["Male Height in Cm"])
min_hombres = min(datos["Male Height in Cm"])
max_hombres = max(datos["Male Height in Cm"])

# Alturas de mujeres.
media_mujeres = mean(datos["Female Height in Cm"])
min_mujeres = min(datos["Female Height in Cm"])
max_mujeres = max(datos["Female Height in Cm"])

print("ALTURA DE HOMBRES")
print(f"Media: {media_hombres:.2f} cm")
print(f"Mínima: {min_hombres:.2f} cm")
print(f"Máxima: {max_hombres:.2f} cm")

print("\nALTURA DE MUJERES")
print(f"Media: {media_mujeres:.2f} cm")
print(f"Mínima: {min_mujeres:.2f} cm")
print(f"Máxima: {max_mujeres:.2f} cm")
```

## Obtener los datos de México
#### Por último, filtro la tabla para encontrar la fila cuyo país sea Mexico, que es como aparece escrito en el CSV.

```python
datos_mexico = datos[datos["Country Name"] == "Mexico"]

print("Datos de México:")
print(datos_mexico.to_string(index=False))
```
