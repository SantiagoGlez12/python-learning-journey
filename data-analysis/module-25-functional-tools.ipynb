## **Ejercicio 1. Función `lambda` para calcular una raíz cuadrada**
### Vamos a calcular la raíz cuadrada de **49**. Primero creamos una función normal con `def` y después hacemos lo mismo con `lambda`.



```python
# Solución sin lambda

def raiz_cuadrada(numero):
    return numero ** 0.5

numero = 49
print("Sin lambda:", raiz_cuadrada(numero))
```

```python
raiz_lambda = lambda numero: numero ** 0.5

numero = 49
print("Con lambda:", raiz_lambda(numero))
```

## **Ejercicio 2. Función `map` para obtener la longitud de cada palabra**
### Partimos de una frase. Con `split()` la dividimos en palabras y con `len()` obtenemos el número de caracteres de cada una.

```python
# Solución sin map

texto = "Python es una herramienta util"
palabras = texto.split()

longitudes_sin_map = []
for palabra in palabras:
    longitudes_sin_map.append(len(palabra))

print("Palabras:", palabras)
print("Sin map:", longitudes_sin_map)
```

```python
# Solución con map

texto = "Python es una herramienta util"
palabras = texto.split()

longitudes_con_map = list(map(len, palabras))

print("Con map:", longitudes_con_map)
```

## **Ejercicio 3. Función `reduce` para calcular el producto de una lista**
### Multiplicaremos los elementos de la lista `[2, 3, 4, 5]`. Para utilizar `reduce` primero debemos importarla de `functools`.



```python
# Solución sin reduce

numeros = [2, 3, 4, 5]
producto_sin_reduce = 1

for numero in numeros:
    producto_sin_reduce = producto_sin_reduce * numero

print("Sin reduce:", producto_sin_reduce)
```

```python
# Solución con reduce

from functools import reduce

numeros = [2, 3, 4, 5]
producto_con_reduce = reduce(lambda a, b: a * b, numeros)

print("Con reduce:", producto_con_reduce)
```

## **Ejercicio 4. Función `filter` para identificar palabras con mayúsculas o números**
### Buscaremos las palabras que contengan **al menos una letra mayúscula o un número**. Primero lo hacemos con ciclos y condicionales; después usamos `filter()`.



```python
# Solución sin filter

palabras = ["Python", "codigo", "Modulo24", "analisis", "Datos", "practica2"]
resultado_sin_filter = []

for palabra in palabras:
    encontrada = False
    for caracter in palabra:
        if caracter.isupper() or caracter.isdigit():
            encontrada = True

    if encontrada:
        resultado_sin_filter.append(palabra)

print("Sin filter:", resultado_sin_filter)
```

```python
palabras = ["Python", "codigo", "Modulo24", "analisis", "Datos", "practica2"]

def contiene_mayuscula_o_numero(palabra):
    for caracter in palabra:
        if caracter.isupper() or caracter.isdigit():
            return True
    return False
resultado_con_filter = list(filter(contiene_mayuscula_o_numero, palabras))

print("Con filter:", resultado_con_filter)
```
